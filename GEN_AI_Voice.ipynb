```python

%%writefile /content/drive/MyDrive/cognitive_mirror/voice/voice_analysis.py

import os
import json
import numpy as np
import pandas as pd
import torch
import librosa
import matplotlib.pyplot as plt
import gradio as gr

from datetime import datetime
from transformers import pipeline

# --------------------------------
# PATHS
# --------------------------------

BASE_DIR = os.path.dirname(os.path.abspath(__file__))

WHISPER_DIR = os.path.join(BASE_DIR, "whisper_model")
EMOTION_DIR = os.path.join(BASE_DIR, "emotion_model")
RESULTS_DIR = os.path.join(BASE_DIR, "voice_results")

os.makedirs(RESULTS_DIR, exist_ok=True)

SUPPORTED_LANGUAGES = {
    "English": "english",
    "Hindi": "hindi",
    "Marathi": "marathi"
}

device = 0 if torch.cuda.is_available() else -1

# --------------------------------
# LOAD MODELS
# --------------------------------

def load_models():

    global asr, emotion_model

    asr = pipeline(
        "automatic-speech-recognition",
        model=WHISPER_DIR if os.path.exists(
            os.path.join(WHISPER_DIR, "config.json")
        ) else "openai/whisper-small",
        device=device
    )

    emotion_model = pipeline(
        "audio-classification",
        model=EMOTION_DIR if os.path.exists(
            os.path.join(EMOTION_DIR, "config.json")
        ) else "superb/wav2vec2-base-superb-er",
        device=device
    )

    print("Voice models loaded successfully!")


# --------------------------------
# AUDIO PREPROCESSING
# --------------------------------

def preprocess_audio(audio_path):

    if audio_path is None:
        raise ValueError("Please upload or record audio.")

    audio, sr = librosa.load(
        audio_path,
        sr=16000,
        mono=True
    )

    if len(audio) == 0:
        raise ValueError("Audio file is empty.")

    if not np.isfinite(audio).all():
        raise ValueError("Audio contains invalid values.")

    duration = len(audio) / sr

    if duration < 1:
        raise ValueError("Audio must be at least 1 second.")

    if np.max(np.abs(audio)) < 0.001:
        raise ValueError("Audio is silent.")

    audio = audio / (np.max(np.abs(audio)) + 1e-8)

    return audio, sr, duration


# --------------------------------
# SPEECH TRANSCRIPTION
# --------------------------------

def transcribe_voice(audio_path, selected_language="English"):

    audio, sr, duration = preprocess_audio(audio_path)

    if selected_language not in SUPPORTED_LANGUAGES:
        raise ValueError("Unsupported language selected.")

    language = SUPPORTED_LANGUAGES[selected_language]

    result = asr(
        {
            "raw": audio,
            "sampling_rate": sr
        },
        generate_kwargs={
            "language": language,
            "task": "transcribe"
        }
    )

    return result["text"].strip()


# --------------------------------
# EMOTION DETECTION
# --------------------------------

def detect_voice_emotion(audio_path):

    audio, sr, duration = preprocess_audio(audio_path)

    predictions = emotion_model(
        {
            "raw": audio,
            "sampling_rate": sr
        }
    )

    df = pd.DataFrame(predictions)

    best = df.iloc[0]

    emotion = best["label"]
    confidence = float(best["score"])

    return emotion, confidence, df


# --------------------------------
# AUDIO FEATURE EXTRACTION
# --------------------------------

def extract_voice_features(audio_path):

    audio, sr, duration = preprocess_audio(audio_path)

    pitch = librosa.yin(
        audio,
        fmin=65,
        fmax=400,
        sr=sr
    )

    rms = librosa.feature.rms(y=audio)[0]

    zcr = librosa.feature.zero_crossing_rate(y=audio)[0]

    mfcc = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=13
    )

    centroid = librosa.feature.spectral_centroid(
        y=audio,
        sr=sr
    )[0]

    features = {
        "Duration (seconds)": round(duration, 2),
        "Average pitch (Hz)": round(float(np.mean(pitch)), 2),
        "Average energy": round(float(np.mean(rms)), 4),
        "Zero crossing rate": round(float(np.mean(zcr)), 4),
        "Spectral centroid": round(float(np.mean(centroid)), 2)
    }

    for i in range(13):
        features[f"MFCC {i+1}"] = round(
            float(np.mean(mfcc[i])), 4
        )

    return pd.DataFrame(
        list(features.items()),
        columns=["Feature", "Value"]
    )


# --------------------------------
# AUDIO GRAPHS
# --------------------------------

def generate_voice_graphs(audio_path):

    audio, sr, duration = preprocess_audio(audio_path)

    time = np.arange(len(audio)) / sr

    pitch = librosa.yin(
        audio,
        fmin=65,
        fmax=400,
        sr=sr
    )

    pitch_time = librosa.times_like(pitch, sr=sr)

    rms = librosa.feature.rms(y=audio)[0]

    rms_time = librosa.frames_to_time(
        np.arange(len(rms)),
        sr=sr
    )

    mfcc = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=13
    )

    fig, axes = plt.subplots(4, 1, figsize=(12, 12))

    axes[0].plot(time, audio)
    axes[0].set_title("Audio Waveform")

    axes[1].plot(pitch_time, pitch)
    axes[1].set_title("Pitch Variation")

    axes[2].plot(rms_time, rms)
    axes[2].set_title("Audio Energy")

    axes[3].imshow(
        mfcc,
        aspect="auto",
        origin="lower"
    )
    axes[3].set_title("MFCC Features")

    plt.tight_layout()

    return fig


# --------------------------------
# COMPLETE VOICE ANALYSIS
# --------------------------------

def analyze_voice(audio_path, selected_language="English"):

    if audio_path is None:
        raise gr.Error("Please upload or record audio.")

    try:

        transcript = transcribe_voice(
            audio_path,
            selected_language
        )

        emotion, confidence, emotion_df = detect_voice_emotion(
            audio_path
        )

        features_df = extract_voice_features(audio_path)

        graph = generate_voice_graphs(audio_path)

        result = {
            "language": selected_language,
            "transcript": transcript,
            "emotion": emotion,
            "confidence": confidence,
            "features": features_df.to_dict(orient="records"),
            "emotion_predictions": emotion_df.to_dict(orient="records")
        }

        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        output_path = os.path.join(
            RESULTS_DIR,
            f"voice_analysis_{timestamp}.json"
        )

        with open(output_path, "w", encoding="utf-8") as f:
            json.dump(
                result,
                f,
                indent=4,
                ensure_ascii=False
            )

        return (
            transcript,
            f"{emotion} ({confidence:.2%})",
            features_df,
            emotion_df,
            graph,
            output_path
        )

    except Exception as e:
        raise gr.Error(str(e))


# --------------------------------
# STANDALONE GRADIO GUI
# --------------------------------

def build_demo():

    with gr.Blocks(title="Cognitive Mirror - Voice Analysis") as demo:

        gr.Markdown("# Cognitive Mirror - Voice Analysis")

        audio_input = gr.Audio(
            sources=["upload", "microphone"],
            type="filepath",
            label="Upload or Record Voice"
        )

        language_input = gr.Dropdown(
            choices=["English", "Hindi", "Marathi"],
            value="Marathi",
            label="Select Voice Language"
        )

        analyze_btn = gr.Button(
            "Analyze Voice",
            variant="primary"
        )

        transcript_output = gr.Textbox(
            label="Speech Transcript",
            lines=4
        )

        emotion_output = gr.Textbox(
            label="Detected Vocal Emotion"
        )

        features_output = gr.Dataframe(
            label="Audio Features"
        )

        emotion_table = gr.Dataframe(
            label="Emotion Predictions"
        )

        graph_output = gr.Plot(
            label="Audio Analysis Graphs"
        )

        file_output = gr.File(
            label="Download Analysis JSON"
        )

        analyze_btn.click(
            fn=analyze_voice,
            inputs=[audio_input, language_input],
            outputs=[
                transcript_output,
                emotion_output,
                features_output,
                emotion_table,
                graph_output,
                file_output
            ]
        )

    return demo


if __name__ == "__main__":

    load_models()

    demo = build_demo()

    demo.launch(share=True)
```
